```python
import torch
import pandas as pd

from transformers import (
    AutoTokenizer,
    AutoModelForSequenceClassification
)
```

```python
train = pd.read_csv("/kaggle/input/datasets/sharankumare/mcq-train/train.csv")

print(train.shape)
train.head()
```

```python
label_map = {
    "A": 0,
    "B": 1,
    "C": 2,
    "D": 3,
    "E": 4
}

train["label"] = train["answer"].map(label_map)

train[["answer", "label"]].head()
```

```python
device = 'cuda'if torch.cuda.is_available() else 'cpu'
device
```

```python
deberta_checkpoint = "microsoft/deberta-v3-small"

deberta_tokenizer = AutoTokenizer.from_pretrained(
    deberta_checkpoint
)

deberta_model = AutoModelForSequenceClassification.from_pretrained(
    deberta_checkpoint,
    num_labels=5
)

deberta_model = deberta_model.to(device)

print("DeBERTa loaded")
```

```python
roberta_checkpoint = "roberta-base"

roberta_tokenizer = AutoTokenizer.from_pretrained(
    roberta_checkpoint
)

roberta_model = AutoModelForSequenceClassification.from_pretrained(
    roberta_checkpoint,
    num_labels=5
)

roberta_model = roberta_model.to(device)

print("RoBERTa loaded")
```

```python
import torch

prompt_25 = str(train.loc[25, "prompt"])

deberta_inputs = deberta_tokenizer(
    prompt_25,
    return_tensors="pt",
    truncation=True,
    padding=True
)

deberta_inputs = {
    key: value.to(device)
    for key, value in deberta_inputs.items()
}

deberta_model.eval()

with torch.no_grad():
    deberta_outputs = deberta_model(**deberta_inputs)

deberta_logits = deberta_outputs.logits

deberta_probs = torch.softmax(
    deberta_logits,
    dim=-1
)

print("Logits:")
print(deberta_logits)

print("\nProbabilities:")
print(deberta_probs)
```

```python
options = ["A", "B", "C", "D", "E"]

highest_label = torch.argmax(
    deberta_probs,
    dim=-1
).item()

highest_probability = deberta_probs[0, highest_label].item()

highest_option = options[highest_label]

print("Option:", highest_option)
print("Probability:", highest_probability)

print(
    f"Answer: {highest_option}, {highest_probability:.4f}"
)
```

```python
# RoBERTa inference on row index 25

prompt_25 = str(train.loc[25, "prompt"])

roberta_inputs = roberta_tokenizer(
    prompt_25,
    return_tensors="pt",
    truncation=True,
    padding=True
)

roberta_inputs = {
    key: value.to(device)
    for key, value in roberta_inputs.items()
}

roberta_model.eval()

with torch.no_grad():
    roberta_outputs = roberta_model(**roberta_inputs)

roberta_probs = torch.softmax(
    roberta_outputs.logits,
    dim=-1
)

print("RoBERTa probabilities:")
print(roberta_probs)
```

```python
average_probs = (
    deberta_probs + roberta_probs
) / 2

options = ["A", "B", "C", "D", "E"]

highest_label = torch.argmax(
    average_probs,
    dim=-1
).item()

highest_option = options[highest_label]

print("DeBERTa probabilities:", deberta_probs)
print("RoBERTa probabilities:", roberta_probs)
print("Average probabilities:", average_probs)

print("Answer:", highest_option)
```

```python
weighted_probs = (
    0.70 * deberta_probs
    + 0.30 * roberta_probs
)

options = ["A", "B", "C", "D", "E"]

highest_label = torch.argmax(
    weighted_probs,
    dim=-1
).item()

highest_option = options[highest_label]

print("DeBERTa probabilities:", deberta_probs)
print("RoBERTa probabilities:", roberta_probs)
print("Weighted probabilities:", weighted_probs)

print("Answer:", highest_option)
```

```python
options = ["A", "B", "C", "D", "E"]

top3_indices = torch.argsort(
    weighted_probs,
    dim=-1,
    descending=True
)[0, :3]

top3_options = [
    options[i.item()]
    for i in top3_indices
]

top3_prediction = " ".join(top3_options)

print("Top-3 indices:", top3_indices)
print("Top-3 options:", top3_options)
print("Kaggle prediction:", top3_prediction)
```

```python
test = pd.read_csv(
    "/kaggle/input/datasets/sharankumare/mcq-test/test.csv"
)

print("Test shape:", test.shape)
test.head()
```

```python
import torch

options = ["A", "B", "C", "D", "E"]

batch_size = 32

predictions = []

deberta_model.eval()
roberta_model.eval()

for start in range(0, len(test), batch_size):

    end = start + batch_size

    prompts = (
        test.iloc[start:end]["prompt"]
        .astype(str)
        .tolist()
    )

    # -------------------------
    # DeBERTa
    # -------------------------

    deberta_inputs = deberta_tokenizer(
        prompts,
        return_tensors="pt",
        padding=True,
        truncation=True
    )

    deberta_inputs = {
        key: value.to(device)
        for key, value in deberta_inputs.items()
    }

    # -------------------------
    # RoBERTa
    # -------------------------

    roberta_inputs = roberta_tokenizer(
        prompts,
        return_tensors="pt",
        padding=True,
        truncation=True
    )

    roberta_inputs = {
        key: value.to(device)
        for key, value in roberta_inputs.items()
    }

    # -------------------------
    # Inference
    # -------------------------

    with torch.no_grad():

        deberta_logits = deberta_model(
            **deberta_inputs
        ).logits

        roberta_logits = roberta_model(
            **roberta_inputs
        ).logits

    # -------------------------
    # Softmax
    # -------------------------

    deberta_probs = torch.softmax(
        deberta_logits,
        dim=-1
    )

    roberta_probs = torch.softmax(
        roberta_logits,
        dim=-1
    )

    # -------------------------
    # Weighted ensemble
    # -------------------------

    final_probs = (
        0.70 * deberta_probs
        + 0.30 * roberta_probs
    )

    # -------------------------
    # Top 3
    # -------------------------

    top3_indices = torch.argsort(
        final_probs,
        dim=-1,
        descending=True
    )[:, :3]

    for indices in top3_indices:

        prediction = " ".join(
            options[i.item()]
            for i in indices
        )

        predictions.append(prediction)

print("Total predictions:", len(predictions))
```

```python
submission = pd.DataFrame({
    "id": test["id"],
    "prediction": predictions
})

submission.to_csv(
    "submission.csv",
    index=False
)

submission.head()
```

```python
import torch

instruction = (
    "Answer the following multiple-choice question carefully:"
)

test_50 = test.iloc[:50].copy()

original_prompts = (
    test_50["prompt"]
    .astype(str)
    .tolist()
)

augmented_prompts = [
    instruction + " " + prompt
    for prompt in original_prompts
]
```

```python
original_inputs = deberta_tokenizer(
    original_prompts,
    return_tensors="pt",
    padding=True,
    truncation=True
)

augmented_inputs = deberta_tokenizer(
    augmented_prompts,
    return_tensors="pt",
    padding=True,
    truncation=True
)

original_inputs = {
    key: value.to(device)
    for key, value in original_inputs.items()
}

augmented_inputs = {
    key: value.to(device)
    for key, value in augmented_inputs.items()
}
```

```python
deberta_model.eval()

with torch.no_grad():

    original_logits = deberta_model(
        **original_inputs
    ).logits

    augmented_logits = deberta_model(
        **augmented_inputs
    ).logits
```

```python
original_probs = torch.softmax(
    original_logits,
    dim=-1
)

augmented_probs = torch.softmax(
    augmented_logits,
    dim=-1
)

tta_probs = (
    original_probs + augmented_probs
) / 2
```

```python
original_top1 = torch.argmax(
    original_probs,
    dim=-1
)

tta_top1 = torch.argmax(
    tta_probs,
    dim=-1
)

different_predictions = (
    original_top1 != tta_top1
).sum().item()

print("Original Top-1:")
print(original_top1)

print("\nTTA Top-1:")
print(tta_top1)

print(
    "\nDifferent Top-1 predictions:",
    different_predictions
)
```

```python
import torch

test_100 = test.iloc[:100].copy()

prompts_100 = (
    test_100["prompt"]
    .astype(str)
    .tolist()
)

# -------------------------
# DeBERTa tokenization
# -------------------------

deberta_inputs_100 = deberta_tokenizer(
    prompts_100,
    return_tensors="pt",
    padding=True,
    truncation=True
)

deberta_inputs_100 = {
    key: value.to(device)
    for key, value in deberta_inputs_100.items()
}

# -------------------------
# RoBERTa tokenization
# -------------------------

roberta_inputs_100 = roberta_tokenizer(
    prompts_100,
    return_tensors="pt",
    padding=True,
    truncation=True
)

roberta_inputs_100 = {
    key: value.to(device)
    for key, value in roberta_inputs_100.items()
}
```

```python
deberta_model.eval()
roberta_model.eval()

with torch.no_grad():

    deberta_logits_100 = deberta_model(
        **deberta_inputs_100
    ).logits

    roberta_logits_100 = roberta_model(
        **roberta_inputs_100
    ).logits
```

```python
deberta_probs_100 = torch.softmax(
    deberta_logits_100,
    dim=-1
)

roberta_probs_100 = torch.softmax(
    roberta_logits_100,
    dim=-1
)
```

```python
weighted_probs_100 = (
    0.70 * deberta_probs_100
    + 0.30 * roberta_probs_100
)
```

```python
deberta_top1 = torch.argmax(
    deberta_probs_100,
    dim=-1
)

ensemble_top1 = torch.argmax(
    weighted_probs_100,
    dim=-1
)

different_count = (
    deberta_top1 != ensemble_top1
).sum().item()

print("DeBERTa Top-1:")
print(deberta_top1)

print("\nEnsemble Top-1:")
print(ensemble_top1)

print(
    "\nDifferent Top-1 predictions:",
    different_count
)
```

```python
# Highest class probability for each row

deberta_confidence = torch.max(
    deberta_probs_100,
    dim=-1
).values

ensemble_confidence = torch.max(
    weighted_probs_100,
    dim=-1
).values

# Confidence Gain
confidence_gain = (
    ensemble_confidence - deberta_confidence
)

# Count strictly positive gains
positive_gain_count = (
    confidence_gain > 0
).sum().item()

print("DeBERTa confidence:")
print(deberta_confidence)

print("\nEnsemble confidence:")
print(ensemble_confidence)

print("\nConfidence gain:")
print(confidence_gain)

print(
    "\nRows with positive confidence gain:",
    positive_gain_count
)
```

```python
# Get ordered Top-3 class indices

deberta_top3 = torch.argsort(
    deberta_probs_100,
    dim=-1,
    descending=True
)[:, :3]

ensemble_top3 = torch.argsort(
    weighted_probs_100,
    dim=-1,
    descending=True
)[:, :3]

# Check whether ANY position changed in each row

top3_changed = (
    deberta_top3 != ensemble_top3
).any(dim=1)

# Count changed rows

top3_changed_count = (
    top3_changed.sum().item()
)

print(
    "Rows with changed ordered Top-3:",
    top3_changed_count
)
```

```python
options = ["A", "B", "C", "D", "E"]

deberta_top3_strings = [
    " ".join(options[i.item()] for i in row)
    for row in deberta_top3
]

ensemble_top3_strings = [
    " ".join(options[i.item()] for i in row)
    for row in ensemble_top3
]

for i in range(100):
    if deberta_top3_strings[i] != ensemble_top3_strings[i]:
        print(
            i,
            "DeBERTa:",
            deberta_top3_strings[i],
            "| Ensemble:",
            ensemble_top3_strings[i]
        )
```

```python
val_100 = train.iloc[:100].copy()

val_prompts = (
    val_100["prompt"]
    .astype(str)
    .tolist()
)
```

```python
deberta_val_inputs = deberta_tokenizer(
    val_prompts,
    return_tensors="pt",
    padding=True,
    truncation=True
)

deberta_val_inputs = {
    key: value.to(device)
    for key, value in deberta_val_inputs.items()
}

roberta_val_inputs = roberta_tokenizer(
    val_prompts,
    return_tensors="pt",
    padding=True,
    truncation=True
)

roberta_val_inputs = {
    key: value.to(device)
    for key, value in roberta_val_inputs.items()
}
```

```python
deberta_model.eval()
roberta_model.eval()

with torch.no_grad():

    deberta_val_logits = deberta_model(
        **deberta_val_inputs
    ).logits

    roberta_val_logits = roberta_model(
        **roberta_val_inputs
    ).logits
```

```python
deberta_val_probs = torch.softmax(
    deberta_val_logits,
    dim=-1
)

roberta_val_probs = torch.softmax(
    roberta_val_logits,
    dim=-1
)

weighted_val_probs = (
    0.70 * deberta_val_probs
    + 0.30 * roberta_val_probs
)
```

```python
top3_predictions = torch.argsort(
    weighted_val_probs,
    dim=-1,
    descending=True
)[:, :3]
```

```python
true_labels = torch.tensor(
    val_100["label"].values,
    dtype=torch.long
)
```

```python
map3_scores = []

for true_label, predictions in zip(
    true_labels,
    top3_predictions.cpu()
):

    predictions = predictions.tolist()
    true_label = true_label.item()

    if true_label == predictions[0]:
        score = 1.0

    elif true_label == predictions[1]:
        score = 1.0 / 2.0

    elif true_label == predictions[2]:
        score = 1.0 / 3.0

    else:
        score = 0.0

    map3_scores.append(score)

map3 = sum(map3_scores) / len(map3_scores)

print("MAP@3:", map3)
print("Rounded MAP@3:", round(map3, 4))
```
